# Exercício 10: Mini pipeline prático

Monte um pequeno roteiro em Python que faça o seguinte:

- leia a base vendas_loja.csv;
- selecione apenas as colunas produto, preco, quantidade e cidade;
- filtre registros com preco maior ou igual a 100;
- filtre apenas registros da cidade de Fortaleza;
- exiba as 10 primeiras linhas do resultado final;
- salve o resultado em um arquivo chamado fortaleza_filtrado.csv.


```python
import pandas as pd

df = pd.read_csv("vendas_loja.csv")

# Seleciona apenas as colunas pedidas
df_pipeline = df[["produto", "preco", "quantidade", "cidade"]]

# Filtra apenas preços maiores ou iguais a 100
df_pipeline = df_pipeline[df_pipeline["preco"] >= 100]

# Filtra apenas a cidade de Fortaleza
df_pipeline = df_pipeline[df_pipeline["cidade"] == "Fortaleza"]

# Mostra as 10 primeiras linhas do resultado final
print(df_pipeline.head(10))

# Salva em um novo arquivo CSV
df_pipeline.to_csv("fortaleza_filtrado.csv", index=False)
```

# Exercício 11: Desafio adicional

Crie um filtro para selecionar apenas vendas da categoria Informática na cidade de Fortaleza.

Depois, calcule a quantidade de linhas desse subconjunto e o preço médio encontrado.


```python
informatica_fortaleza = df[(df["categoria"] == "Informática") & (df["cidade"] == "Fortaleza")]
quantidade_linhas = len(informatica_fortaleza)
preco_medio_info_fortal = informatica_fortaleza["preco"].mean()

print("Quantidade de linhas do filtro:", quantidade_linhas)
print("Preço médio do filtro:", preco_medio_info_fortal)
```
