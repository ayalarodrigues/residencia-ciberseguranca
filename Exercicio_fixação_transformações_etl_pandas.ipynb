```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'cliente': ['Ana', 'Bruno', 'CARLA', None],
    'idade': [25, None, 31, 40],
    'cidade': ['Fortaleza', 'fortaleza', 'Recife', ''],
    'valor_compra': ['120,50', '89,90', None, '210,00'],
               'data': ['2026-03-01', '01/03/2026', '2026/03/02', None]

})

print(df)
```

Exercício de fixação

1. Quais colunas têm maior risco de erro;

**R: As colunas com maior risco de erro são 'cliente, 'cidade', 'valor_compra' e 'data', pois elas não estão padronizadas, possuem valores ausentes, formatos diferentes  e campos vazios. A coluna idade também precisa de atenção por possuir valor nulo.**

2. Onde faz sentido preencher nulos;

**R: Faz sentido preencher nulos principalmente em 'idade', usando média ou mediana. Também pode ser adequado preencher 'cliente' e 'cidade' com algum valor padrão, como 'Desconhecido'ou 'Não informado'. Já em 'data' e 'valor_compra', o preenchimento é mais complicado porque envolve dados mais sensíveis.**

3. Onde faz sentido remover registros.

**R: Faz sentido remover registros principalmente quando faltam dados importantes, como 'valor_compra' e 'data'. Registros também podem ser removidos caso existam muitos erros. Porém, alguns nulos podem ser tratados sem a exclusão do registro, como acontece em 'idade'.**

```python
df.info()
```

```python
#Verifica valores vazios em cidade

print(df['cidade'] == "")
```

```python
#Preenchimento de nulos

#Preenche idade com mediana

df["idade"] = df["idade"].fillna(df["idade"].median())
df
```

```python
#Preenche cliente com um texto padrão

df["cliente"] = df["cliente"].fillna("Desconhecido")
df
```

```python
# Preenche cidade vazia com 'Não informado"

df["cidade"] = df["cidade"].replace("", "Não informado")
df
```

```python
# Padronização de nomes dos clientes

#Coloca a primeira letra da cada palavra em maiúscula e o resto em minúscula

df["cliente"] = df["cliente"].str.title()
df
```

```python
# Padronização de cidades

df["cidade"] = df["cidade"].str.title()
df
```

```python
# Converter 'valor_compra' para número

df["valor_compra"] = df["valor_compra"].str.replace(",", ".", regex=False) #troca a vírgula por ponto
df["valor_compra"] = pd.to_numeric(df["valor_compra"], errors="coerce") # 'coerce' transforma inválido em nulo

df['valor_compra']
```

```python
#Converte 'data' para formato de data

df["data"] = pd.to_datetime(df["data"], errors="coerce", dayfirst=True)
df["data"]
```

```python
#Remove registros com 'valor_compra' nulo

df_sem_valor_nulo = df.dropna(subset=["valor_compra"])
df_sem_valor_nulo
```
