Exercício 4: Leitura e transformação simples

Suponha o seguinte arquivo alunos.csv:

nome,nota
Ana,8.0
Bruno,6.5
Carla,9.0
Diego,7.5

10. Escreva um código em Python que leia o arquivo.
11. Ignore o cabeçalho.
12. Armazene as notas em uma lista.
13. Calcule a média das notas.

```python
with open("alunos.csv", "w", encoding="utf-8") as arquivo:
  arquivo.write("nome,nota\n")
  arquivo.write("Ana,8.0\n")
  arquivo.write("Bruno,6.5\n")
  arquivo.write("Carla,9.0\n")
  arquivo.write("Diego,7.5\n")
```

```python
notas = [] #lista vazia para armazenar notas

with open("alunos.csv", "r", encoding="utf-8") as arquivo:
  linhas = arquivo.readlines()
  for linha in linhas[1:]: #percorre as linhas a partir da posição 1, ignorando a primeira linha, que é o cabeçalho
    partes = linha.strip().split(",") #strip remove espaços e o \n e o split divide a linha em parte usando vírgula
    nota = float(partes[1]) #pega a segunda parte da linha, que é a nota, e converte para decimal
    notas.append(nota) #adiciona um elemento ao final da lista e vai modificando

media = sum(notas) / len(notas)

print("Lista de notas:", notas)
print("Média de Notas:", media)
```

16. Escreva um pequeno código que percorra a lista e imprima cada nome e nota separadamente.

```python
dados = [
    "Ana,8.0",
    "Bruno,6.5",
    "Carla,9.0",
    "Bruno,6.5"
]

for item in dados:
  partes = item.split(",")
  nome = partes[0]
  nota = partes[1]
  print("Nome:", nome)
  print("Nota:", nota)
```
